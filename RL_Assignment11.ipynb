```python
# Policy Iteration in a Deterministic 2x2 Grid World

# Define states and actions
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

# Initial policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Initialize state values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Grid positions:
# S0  W
# S1  G

positions = {
    'S0': (0, 0),
    'W': (0, 1),
    'S1': (1, 0),
    'G': (1, 1)
}

# Define action movements
moves = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

# Find next state and reward
def get_next_state(state, action):

    if state == 'G':
        return 'G', 0

    r, c = positions[state]
    dr, dc = moves[action]

    new_pos = (r + dr, c + dc)

    # Check boundaries and wall
    if new_pos not in positions.values() or new_pos == positions['W']:
        return state, -1

    # Check goal
    if new_pos == positions['G']:
        return 'G', 100

    # Normal movement
    for s, pos in positions.items():
        if pos == new_pos:
            return s, -1


# Policy Evaluation
def policy_evaluation(policy, V):

    iteration = 0

    while True:
        delta = 0

        for state in ['S0', 'S1']:

            old_value = V[state]

            next_state, reward = get_next_state(
                state, policy[state]
            )

            V[state] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[state]))

        iteration += 1

        print(f"Evaluation {iteration}: {V}")

        if delta < theta:
            break

    return V


# Calculate action values
def calculate_q(state, action, V):

    next_state, reward = get_next_state(state, action)

    return reward + gamma * V[next_state]


# Policy Improvement
def policy_improvement(policy, V):

    policy_stable = True

    for state in ['S0', 'S1']:

        old_action = policy[state]

        q_values = {}

        for action in actions:
            q_values[action] = calculate_q(
                state, action, V
            )

        best_action = max(q_values, key=q_values.get)

        policy[state] = best_action

        if old_action != best_action:
            policy_stable = False

        print(f"\nState: {state}")
        print("Action Values:", q_values)
        print("Best Action:", best_action)

    return policy, policy_stable


# Policy Iteration
def policy_iteration():

    global V, policy

    iteration = 0

    while True:

        iteration += 1

        print("\n========== POLICY ITERATION", iteration, "==========")

        # Step 1: Policy Evaluation
        V = policy_evaluation(policy, V)

        # Step 2: Policy Improvement
        policy, stable = policy_improvement(policy, V)

        print("\nUpdated Policy:", policy)

        # Step 3: Check policy stability
        if stable:
            break

    print("\n========== FINAL RESULTS ==========")

    print("\nOptimal State Values:")
    for state in states:
        print(state, ":", V[state])

    print("\nOptimal Policy:")
    for state in ['S0', 'S1']:
        print(state, ":", policy[state])

    print("\nFinal Action Values:")
    for state in ['S0', 'S1']:
        for action in actions:
            q = calculate_q(state, action, V)
            print(state, action, ":", round(q, 2))

    print("\nBellman Optimality Verification:")
    for state in ['S0', 'S1']:
        max_q = max(
            calculate_q(state, action, V)
            for action in actions
        )
        print(
            state,
            "V* =", round(V[state], 2),
            "Max Q* =", round(max_q, 2)
        )


# Execute program
policy_iteration()
```
