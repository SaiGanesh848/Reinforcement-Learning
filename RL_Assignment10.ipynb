```python
# Practical 10: Bellman Optimality and Value Iteration
# Deterministic Grid World

# 1. Define states, actions and parameters

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

normal_reward = -1
wall_reward = -1
goal_reward = 10

# Transition mapping
# Grid:
# S0 | Wall
# S1 | G

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },
    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}

# 2. Inspect the MDP

print("STATES:", states)
print("ACTIONS:", actions)
print("Discount Factor:", gamma)
print("Normal Reward:", normal_reward)
print("Wall Reward:", wall_reward)
print("Goal Reward:", goal_reward)

print("\nTRANSITION MAPPING")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, "--", action, "-->", next_state,
              "Reward:", reward)

# 3. Initialize state values

V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

print("\nINITIAL STATE VALUES")
print(V)

# 4. Implement Value Iteration

iteration = 0

print("\nVALUE ITERATION")

while True:
    iteration += 1

    # Synchronous update
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    # Calculate maximum change
    delta = max(
        abs(new_V[state] - V[state])
        for state in ['S0', 'S1']
    )

    V = new_V

    print(f"Iteration {iteration}: "
          f"S0 = {V['S0']:.4f}, "
          f"S1 = {V['S1']:.4f}, "
          f"G = {V['G']:.4f}, "
          f"Delta = {delta:.6f}")

    if delta < theta:
        break

# 5. Display final optimal state values

print("\nFINAL OPTIMAL STATE VALUES")

for state in states:
    print(state, "=", round(V[state], 4))

# 6. Calculate action values

print("\nACTION VALUES Q*(s,a)")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        q = reward + gamma * V[next_state]
        Q[state][action] = q

        print(f"Q({state}, {action}) = {q:.4f}")

# 7. Extract optimal policy

print("\nOPTIMAL POLICY")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    best_actions = [
        action for action, value in Q[state].items()
        if abs(value - max_q) < 1e-9
    ]

    print(state, ":", best_actions)

# 8. Verify Bellman Optimality

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    print(
        f"State: {state}, "
        f"V*(s) = {V[state]:.4f}, "
        f"Max Q*(s,a) = {max_q:.4f}"
    )

    if abs(V[state] - max_q) < 1e-9:
        print("Verification Successful")
    else:
        print("Verification Failed")
```
