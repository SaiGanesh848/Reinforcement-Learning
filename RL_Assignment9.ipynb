```python
gamma = 0.9

# Solve equations algebraically
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Verify state values
check_S0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_S1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("Assignment 1 Results")
print("--------------------")
print("V(S0):", round(V_S0, 2))
print("V(S1):", round(V_S1, 2))

print("\nAction Values")
print("Q(S0, Down):", round(Q_S0_Down, 2))
print("Q(S0, Right):", round(Q_S0_Right, 2))
print("Q(S1, Right):", round(Q_S1_Right, 2))
print("Q(S1, Up):", round(Q_S1_Up, 2))

print("\nVerification")
print("Verified V(S0):", round(check_S0, 2))
print("Verified V(S1):", round(check_S1, 2))
```

```python
gamma = 0.9
theta = 0.000001

# Initialize state values
V_S0 = 0.0
V_S1 = 0.0

iterations = 0

# Iterative Policy Evaluation
while True:

    # Calculate new state values using old values
    new_V_S0 = -1 + 0.81 * V_S1
    new_V_S1 = 89.9 + 0.09 * V_S0

    # Calculate maximum change
    delta = max(
        abs(new_V_S0 - V_S0),
        abs(new_V_S1 - V_S1)
    )

    # Update state values
    V_S0 = new_V_S0
    V_S1 = new_V_S1

    iterations += 1

    # Check convergence
    if delta < theta:
        break

# Calculate action values
Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V_S1)

Q_S0_Right = -1

Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V_S0)

Q_S1_Up = -1 + gamma * V_S0

# Display results
print("Assignment 2 Results")
print("--------------------")

print("Number of iterations:", iterations)

print("\nState Values")
print("V(S0):", round(V_S0, 2))
print("V(S1):", round(V_S1, 2))

print("\nAction Values")
print("Q(S0, Down):", round(Q_S0_Down, 2))
print("Q(S0, Right):", round(Q_S0_Right, 2))
print("Q(S1, Right):", round(Q_S1_Right, 2))
print("Q(S1, Up):", round(Q_S1_Up, 2))

# Policy consistency checks
print("\nPolicy Consistency Checks")

print("V(S0) = Q(S0, Down):",
      abs(V_S0 - Q_S0_Down) < theta)

print("V(S1) = Q(S1, Right):",
      abs(V_S1 - Q_S1_Right) < theta)
```
